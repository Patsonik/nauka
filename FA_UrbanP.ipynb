## Loops and Conditions

```python
# Using of a "for loop" and print every room
print("\nRoom List with Labels:")
for index, room in enumerate(room_names):
    print(f"Room {index+1}: {room}")          # index starts at 0, annd it adds +1 to display room numbers starting from 1


# Using of a "for loop" and print every room without "offices" +print"/n" to separate and be more redable next lines
print("\nRoom List with Labels (excluding all Offices):")
for index, room in enumerate(room_names):     # Loop through the room_names list again using enumerate
    if not room.lower().startswith("office"): # Check if the room name does NOT start with "office" (additional by me)
        print(f"Room {index+1}: {room}")      # If condition is True, print the room with its number
```

```python
# While loop reaches a total area of 120 sqm adding 30 sqm, starting from 0
adding_area = 30              # Set how much area will be added in each loop iteration
total_area = 0                # Starting point = 0
room_count = 0                # To count how many rooms were added

while total_area < 120:
  total_area += adding_area   # Every time add 30 sqm to the total area
  room_count += 1             # Every time increase the room number by 1

print(f"\nAdding area for {room_count} rooms. Total area {total_area} sqm ")
```

## Define and use class

```python
# Definition of a room class
class Room:
    def __init__(self, name, area, use):
      # Initialize the room's name, area, and use
        self.name = name
        self.area = area
        self.use = use
    def describe(self):
      print("Room name is", self.name, "and area of this room is", self.area, "sqm. This room is using as a ", self.use)
```

```python
# Create a room objects
room1 = Room("Break room", 25, "Relaxation and to get your thoughts together")
room2 = Room("Technical room", 35, "Work at technical things")

# Call the describe method
room1.describe()
room2.describe()

```

### Nested data structure

```python
# Nested data structure for rooms grouped by floors
building_floors = {
    "Ground floor": [
        { "name": "Office A", "area": 40, "type": "astronaut research control room"},
        { "name": "Exercise room", "area": 40, "type": "astronaut training room"},

    ],
    "First floor":[
        { "name": "Office B", "area": 30, "type": "Astronaut office room"},
        { "name": "kitchen", "area": 20, "type": "astronaut training room"},
        { "name": "Office C", "area": 30, "type": "Astronaut office room"},

    ],
    "Second floor":[
        { "name": "Toilet1", "area": 25, "type": "Womens toilet"},
        { "name": "Toilet2", "area": 25, "type": "Men toilet"},
        { "name": "Chilloom", "area": 30, "type": "Astronaut rest"},

    ],
    "Third floor":[
        { "name": "Office D", "area": 80, "type": "Astronaut office room"},
    ],

    "Fourth floor":[
        { "name": "Conference room", "area": 80, "type": "Astronaut conference room"},
    ],
}

# Print each floor with theirs names and types
print("Nested building structure:")
for floor, rooms in building_floors.items():        # Loop through each floor and its list of rooms
  print("Floor:")                                   # Print the floor name
  for room in rooms:                                # Loop through each room dictionary in the floor
    print(f"  - {room['name']} ({room['area']} m^2, {room['type']})")

```
